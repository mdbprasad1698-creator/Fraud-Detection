## 9. Key EDA Findings

| Finding | Insight |
|---|---|
| **Fraud rate** | ~9.17% – moderately imbalanced |
| **Amount** | Fraud transactions skew higher |
| **Night hours** | Higher fraud rate between 0–5 AM |
| **KYC** | Unverified users have ~3× higher fraud rate |
| **OTP** | Transactions without OTP have higher fraud |
| **New accounts** | Short account age correlates with fraud |
